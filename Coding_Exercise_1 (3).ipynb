### **Dataset 1: County Credit Card Delinquency**

This dataset provides county-level information about credit card delinquency rates in 2024 and changes in rates from 2022 to 2024. It also includes the estimated number of county residents with credit card delinquencies in 2024.

**Features**

**FIPS:**
Geographic codes used to identify each county

**County:**
The names of each county used in the dataset

**State:**
The state where each county is located

**Credit Card Delinquency Rate, 2024:**
The percentage of credit cards that were delinquent in 2024

**Percent Change in Credit Card Delinquency Rate, 2022 to 2024:**
The percentage change in the county's credit card delinquency rate between the years 2022 and 2024  

**Number of Residents with a Credit Card Delinquency, 2024:**
The estimated number of residents in the county with credit card delinquency in 2024

## **Dataset 2 Gas Prices Combined**

This dataset provides state-level information on the real and nominal gas prices between the years 2024 and 2026. It includes geographic information for each state along with nominal and inflation adjusted gas prices recorded at specific dates.

**Features**

**State_name:** The name of the state where the gas price data was collected

**State_abbr:**
Two-letter abbreviation used to identify the state

**State_code:**
Numerical code assigned to each state

**stcd:**
Geographic code used to identify each state

**cd_code:**
Geographic code used to identify the congressional district

**date:**
Calendar date  when the price was recorded

**gasprices_nominal:**
Gas prices based on the wholesale value at the time of data collection

**gasprices_real:**
The gas price adjusted for inflation, to allow for comparison across different time periods

## **Dataset 3: Community College**

The dataset provides information on estimated tuition and funding cost associated with providing free community college. It includes tuition and fees, federal and state payments, total state costs, and excess funding.

**Features**

**State1:** Name of the state represented in the dataset

**Average Tuition and Fees:** Average amount students would pay for tuition and fees at community college

**Federal Payment:** Amount of funding provided through federal financial support

**20% State Payment:** The estimated amount the state would contribute based on 20% of tuition and fees

**Additional State Payment:** The additional funding that would be provided by the state

**Total State Cost:** Total estimated cost to the state after accounting for applicable payment and funding

**Excess Funding:** Amount of funding leftover after estimated costs are covered
